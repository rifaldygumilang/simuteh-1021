```python
import cv2import numpy as npdef jalankan_morfologi_bercak(biner):    # Operasi Morfologi Opening dan Closing untuk analisis bercak daun    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))    opening = cv2.morphologyEx(biner, cv2.MORPH_OPEN, kernel)    closing = cv2.morphologyEx(opening, cv2.MORPH_CLOSE, kernel, iterations=2)    return closingprint("Modul P6 Morfologi & Analisis Bercak siap.")
```
